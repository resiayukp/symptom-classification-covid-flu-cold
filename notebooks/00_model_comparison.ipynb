```python
#Define features for X array and label for y array
features = df.columns[:20]
label = ['TYPE']
```

```python
#Define X and y
X = df[features]
y = df[label]
```

```python
#Define function to change y data type from Obj to Int
from collections import OrderedDict
def replace_all(text, dic):
    for i, j in dic.items():
        text = text.replace(i, j)
    return text
od = OrderedDict([("ALLERGY", "0"), ("COLD", "1"),("COVID", "2"), ("FLU", "3")])
```

```python
#Import library to split the data for train and test
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, hamming_loss, precision_score, recall_score, f1_score, roc_auc_score

#Set data size for test to 20%
train, test = train_test_split(df, test_size=0.2, random_state=10, shuffle = True, stratify = y) #Set stratify to handle imbalanced dataset
X_train, y_train = train[features],train[label]
X_test, y_test = test[features],test[label]

#Change y data dype
y_train = replace_all(y_train, od).astype(int).values.ravel()
y_test = replace_all(y_test, od).astype(int).values.ravel()

#Print data dimension
print('training dataset')
print(X_train.shape)
print(y_train.shape)
print()
print('testing dataset:')
print(X_test.shape)
print(y_test.shape)
print(y_test)
```

```python
import numpy as np

# Check for highly correlated feature pairs
corr_matrix = X_train.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Flag features with correlation > 0.8 to another feature
to_drop = [col for col in upper.columns if any(upper[col] > 0.8)]
print("Highly correlated columns to consider dropping:", to_drop)

X_train_reduced = X_train.drop(columns=to_drop)
X_test_reduced = X_test.drop(columns=to_drop)
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

#Was using k=10 here before - tested it against keeping all 20 and it tanks recall on
#COLD/COVID (the small classes) because it throws away symptoms that only matter for those.
#Only 20 features total so there's no real need to cut any of them, just keeping the
#score plot to see what's driving the predictions.
selector = SelectKBest(score_func=f_classif, k='all')

X_train_selected = selector.fit_transform(X_train_reduced, y_train)
X_test_selected = selector.transform(X_test_reduced)

print(f"Original number of features: {X_train.shape[1]}")
print(f"Features used: {X_train_selected.shape[1]}\n")

feature_scores = pd.DataFrame({
    'Feature': X_train_reduced.columns,
    'Score': selector.scores_
}).sort_values(by='Score', ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x='Score', y='Feature', hue='Feature', data=feature_scores.head(15), palette='viridis', legend=False)
plt.title('Top Features by Importance (ANOVA F-value)')
plt.xlabel('Importance Score')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()
```

Support Vector Classifier

```python
from sklearn.model_selection import GridSearchCV
from datetime import datetime
from sklearn.svm import SVC

start = datetime.now()

#class_weight='balanced' matters a lot here - COVID and COLD are under 5% of the data combined,
#without it the model barely bothers learning them. Trimmed the grid a bit (gamma='auto' almost
#never won) and dropped cv to 3 since 5-fold was taking forever on 44k rows.
svc = SVC(random_state=42, class_weight='balanced')

param_grid_svc = {
    'C': [0.1, 1, 10],
    'kernel': ['linear', 'rbf'],
    'gamma': ['scale']
}

grid_search_svc = GridSearchCV(
    estimator=svc,
    param_grid=param_grid_svc,
    cv=3,
    scoring='recall_macro',
    n_jobs=-1
)

grid_search_svc.fit(X_train_selected, y_train)

best_svc = grid_search_svc.best_estimator_
print(f"Optimal SVC Hyperparameters: {grid_search_svc.best_params_}")
print('Run time : ', datetime.now() - start)
```

Gradient Boosting (GBM)

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.utils.class_weight import compute_sample_weight

#GBM (and XGBoost/MLP below) don't have class_weight, so the same balancing is done with
#sample weights: rows from COLD/COVID get a bigger weight than FLU rows.
#Without this GBM only gets ~0.75 balanced accuracy, with it ~0.96
sw_train = compute_sample_weight('balanced', y_train)

gbm = GradientBoostingClassifier(random_state=42)

param_grid_gbm = {
    'n_estimators': [100, 200],
    'max_depth': [3, 5],
    'learning_rate': [0.05, 0.1],
    'subsample': [0.8, 1.0]
}

grid_search_gbm = GridSearchCV(
    estimator=gbm,
    param_grid=param_grid_gbm,
    cv=3,
    scoring='recall_macro',
    n_jobs=-1
)

grid_search_gbm.fit(X_train_selected, y_train, sample_weight=sw_train)

best_gbm = grid_search_gbm.best_estimator_
print(f"Optimal GBM Hyperparameters: {grid_search_gbm.best_params_}")
```

Neural Network (MLP)

```python
from sklearn.neural_network import MLPClassifier

#Older scikit-learn versions (e.g. the one in Colab) don't let MLPClassifier.fit() take
#sample_weight, which caused the "unexpected keyword argument 'sample_weight'" error.
#This small subclass accepts it anyway: it re-draws the training rows with probability
#proportional to the weight, so COLD/COVID rows show up as often as FLU rows.
class WeightedMLP(MLPClassifier):
    def fit(self, X, y, sample_weight=None):
        if sample_weight is not None:
            X = np.asarray(X)
            y = np.asarray(y).ravel()
            rng = np.random.RandomState(self.random_state)
            p = np.asarray(sample_weight, dtype=float)
            idx = rng.choice(len(y), size=len(y), replace=True, p=p / p.sum())
            X, y = X[idx], y[idx]
        return super().fit(X, y)

#early_stopping keeps 10% of the training data aside and stops when it stops improving
mlp = WeightedMLP(early_stopping=True, max_iter=300, random_state=42)

param_grid_mlp = {
    'hidden_layer_sizes': [(32,), (64, 32)],
    'alpha': [0.0001, 0.01],
    'learning_rate_init': [0.001, 0.005]
}

grid_search_mlp = GridSearchCV(
    estimator=mlp,
    param_grid=param_grid_mlp,
    cv=3,
    scoring='recall_macro',
    n_jobs=-1
)

grid_search_mlp.fit(X_train_selected, y_train, sample_weight=sw_train)

best_mlp = grid_search_mlp.best_estimator_
print(f"Optimal Neural Network Hyperparameters: {grid_search_mlp.best_params_}")
```

```python
import numpy as np

def calculate_specificity(y_true, y_pred, n_classes):
    """
    Calculates Specificity (True Negative Rate) for binary and multi-class target labels.
    Specificity = TN / (TN + FP)
    """
    cm = confusion_matrix(y_true, y_pred)
    if n_classes == 2:
        tn, fp, fn, tp = cm.ravel()
        return tn / (tn + fp) if (tn + fp) > 0 else 0
    else:
        # Multi-class One-vs-Rest Specificity (Macro Average)
        specificities = []
        for i in range(n_classes):
            tp = cm[i, i]
            fp = cm[:, i].sum() - tp
            fn = cm[i, :].sum() - tp
            tn = cm.sum() - (tp + fp + fn)
            spec = tn / (tn + fp) if (tn + fp) > 0 else 0
            specificities.append(spec)
        return np.mean(specificities)
```

```python
models = {
    'KNN': best_knn,
    'LR': best_lr,
    'SVC': best_svc,
    'DT': best_dt,
    'RF': best_rf,
    'GBM': best_gbm,
    'XGBoost': best_xgb,
    'MLP': best_mlp
}

#predicting once and reusing below instead of calling .predict in every single cell
#(SVC predict isn't free on 9k test rows, no reason to do it 8 times)
preds = {name: m.predict(X_test_selected) for name, m in models.items()}

result = []
for name, y_pred in preds.items():
    result.append({
        "Algorithm": name,
        "Precision": precision_score(y_test, y_pred, average = 'macro', zero_division=0),
        "Recall": recall_score(y_test, y_pred, average = 'macro', zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, average = 'macro', zero_division=0),
        "Specificity": calculate_specificity(y_test, y_pred, 4)
    })
```

Accuracy and balanced accuracy

```python
from sklearn.metrics import balanced_accuracy_score

#accuracy alone hides the small classes (COLD and COVID are ~7% of the data), so also look at
#balanced accuracy = average recall over the 4 classes
accuracy_scores = [accuracy_score(y_test, y_pred) for y_pred in preds.values()]
balanced_scores = [balanced_accuracy_score(y_test, y_pred) for y_pred in preds.values()]
print(accuracy_scores)
print(balanced_scores)
```

```python
#Plot accuracy and balanced accuracy side by side
x = np.arange(len(preds))
plt.bar(x - 0.2, accuracy_scores, width=0.4, label='Accuracy')
plt.bar(x + 0.2, balanced_scores, width=0.4, label='Balanced accuracy')
plt.xticks(x, preds.keys(), rotation=20)
plt.ylim(0, 1.05)
plt.title('Accuracy vs balanced accuracy', fontsize=15, color='r')
plt.ylabel('Score', fontsize=14, color='g')
plt.legend()
plt.tight_layout()
plt.show()
```
